```python

import pandas as pd

# 1. Load the dataset
df = pd.read_csv("cleaned_city.csv")

# Convert Datetime column into date format
df["Datetime"] = pd.to_datetime(df["Datetime"])

# 2. Display first 5 rows
print("FIRST 5 ROWS:")
print(df.head())

# 3. Display last 5 rows
print("\nLAST 5 ROWS:")
print(df.tail())

# 4. Display dataset shape
print("\nDATASET SHAPE:")
print(df.shape)

# 5. Display column names
print("\nCOLUMN NAMES:")
print(df.columns)

# 6. Display data types
print("\nDATA TYPES:")
print(df.dtypes)

# 7. Statistical summary
print("\nSTATISTICAL SUMMARY:")
print(df.describe())

# 8. Find unique cities
print("\nUNIQUE CITIES:")
print(df["City"].unique())

# 9. Average AQI of each city
print("\nAVERAGE AQI BY CITY:")
city_aqi = df.groupby("City")["AQI"].mean()
print(city_aqi)

# 10. Average pollutant levels
print("\nAVERAGE POLLUTANT LEVELS:")
pollutants = ["PM2.5", "PM10", "NO", "NO2",
              "NOx", "NH3", "CO", "SO2", "O3"]

print(df[pollutants].mean())

# 11. Count AQI categories
print("\nAQI CATEGORY COUNTS:")
print(df["AQI_Bucket"].value_counts())

# 12. AQI category count by city
print("\nAQI CATEGORY BY CITY:")
print(pd.crosstab(df["City"], df["AQI_Bucket"]))

# 13. Sort data by AQI
print("\nTOP 10 HIGHEST AQI RECORDS:")
print(df.sort_values(by="AQI", ascending=False).head(10))

# 14. Filter records with Severe AQI
print("\nSEVERE AQI RECORDS:")
severe = df[df["AQI_Bucket"] == "Severe"]
print(severe.head(10))

# 15. Find maximum AQI in each city
print("\nMAXIMUM AQI BY CITY:")
print(df.groupby("City")["AQI"].max())

# 16. Find minimum AQI in each city
print("\nMINIMUM AQI BY CITY:")
print(df.groupby("City")["AQI"].min())

# 17. Year-wise average AQI
df["Year"] = df["Datetime"].dt.year

print("\nYEAR-WISE AVERAGE AQI:")
print(df.groupby("Year")["AQI"].mean())

# 18. Month-wise average AQI
df["Month"] = df["Datetime"].dt.month

print("\nMONTH-WISE AVERAGE AQI:")
print(df.groupby("Month")["AQI"].mean())

# 19. City-wise pollutant analysis
print("\nCITY-WISE AVERAGE POLLUTANTS:")
print(df.groupby("City")[pollutants].mean())


```
